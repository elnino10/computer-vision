```python
import cv2 as cv
import numpy as np
import time as t
```

```python
camera_src = 0

while (True):
    cap = cv.VideoCapture(camera_src)
    if cap.isOpened() == False:
        print("Unable to read camera feed")
        break

    frame_width = cap.get(cv.CAP_PROP_FRAME_WIDTH)
    frame_height = cap.get(cv.CAP_PROP_FRAME_HEIGHT)
    frames_per_sec = cap.get(cv.CAP_PROP_FPS)

    num_frames = 100
    start_time = t.time()
    for i in range(0, num_frames):
        if i % 10 == 0:
            print(f"frame_number: {i}")
        res, frame = cap.read()

    end_time = t.time()
    time_elapsed = end_time - start_time # time is seconds
    fps = num_frames / time_elapsed
    print(f"Camera no: {camera_src}\nWidth: {frame_width}\nHeight: {frame_height}\nFPS: {fps}\nTime: {time_elapsed}")
    camera_src += 1
    cap.release()
```
