```python
import cv2 as cv
import matplotlib.pyplot as plt

image_path = '../../cv-images/road-image.jpg'
save_path = '../../cv-images/saved-image.jpg'
```

![road-image](../../cv-images/road-image.jpg)

```python
image = cv.imread(image_path)
img2 = cv.imread(image_path, cv.IMREAD_GRAYSCALE)
h, w = image.shape[:2]
print(f"Height: {h}, Width: {w}")
```

```python
if img2 is None:
    print("Image not found")
    exit()
else:
    cv.imwrite(save_path, img2)
```

```python
(B, G, R) = image[100, 100]
print(f"Red = {R}\nGreen = {G}\nBlue = {B}")

print('---------------------')
_R = image[100, 100, 2]
print(f"The saturation of red in row 100 column 100 is {_R}")
```

```python
# Get region of interest between 500 -> 699 px on 'y' and 200 -> 499 on 'x'
if image is None:
    print("Image not found")
    exit()
else:
    roi = image[500:700, 200:500]
    cv.imshow("Region of Interest", roi)
    k = cv.waitKey(0) & 0xFF

    if k == 27:
        cv.destroyWindow("Region of Interest")
    elif k == ord('d'):
        cv.destroyAllWindows()
```

```python
# Resize image
ratio = 800 / w
dimension = (800, int(h * ratio))
resize_aspect = cv.resize(img2, dimension)
cv.imshow("Resized Image", resize_aspect)
cv.waitKey(2000)
cv.destroyWindow("Resized Image")
```

```python
# Draw a rectangular shape in the image
output = image.copy()
spot_area = cv.rectangle(output, (350, 430), (435, 500), (255, 0, 255), 2)
cv.imshow("Spot area", spot_area)
cv.waitKey(2000)
```

```python
# Displaying text
image_cpy = image.copy()
text = cv.putText(image_cpy, "Sample Text", (250, 200), cv.FONT_HERSHEY_COMPLEX, 1, (255, 0, 0), 1)
cv.imshow("Image With Text", text)
cv.waitKey(2000)
cv.destroyWindow("Image With Text")
```
