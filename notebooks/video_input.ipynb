```python
import cv2 as cv


video_path = "../../cv-files/class-video.mp4"
cap = cv.VideoCapture(video_path)
```

```python
while (cap.isOpened()):
    res, frame = cap.read()

    if res == False:
        break
    cv.imshow('New Image', frame)
    k = cv.waitKey(25) & 0xFF
    if k == 27:
        break
cv.destroyAllWindows()
cap.release()
```
